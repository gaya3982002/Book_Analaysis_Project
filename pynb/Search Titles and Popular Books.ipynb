### 1.0. Getting "BookTitle" from Partial Name

- Not everyone can remember the exact name of book nor they know what was listed name of that book in our dataset so we are going to use the function which can search the name of books which contain the partial name in it.

```python
# List of all book Titles
all_books_titles = list(books.BookTitle.values)
```

```python
# Function to get Books containing partial name

def BookTitles_from_partialTitle(partial):
    for name in all_books_titles:
        if partial in name:
            print(name)
```

### 2.0 Get Top 20 Recommended books based on popularity

- In **popularity based approach** of recommendation system, we can check about the books which are in trend or are most popular among the users and directly recommend them.
- We will be showing Top 20 Recommended books based on popularity index which is calcuted using Weighted Rating formula

- WR = [(v * R)/(v + m)] + [(m * c)/(v + m)]
    
    where,
    v is the number of votes for the books;
    
    m is the minimum votes required to be listed in the chart;
    
    R is the average rating of the book; and
    
    C is the mean vote across the whole report.

```python
ratings = books["AverageRating"]
rating_count = books["TotalRatingCount"]
MR = ratings.convert_dtypes(np.float16).mean()
print("Mean Rating (MR) accross all books: ", MR)

q = rating_count.quantile(0.99)
print("Minimum number of required ratings (q): ", q)
```

```python
# our weighted rating algorithm
def weighted_rating(books):
    v = books['TotalRatingCount']
    R = books['AverageRating']
    return (v/(v+q) * R) + (q/(q+v) * MR)
```

```python
books.shape
```

```python
top_books = books[(books["TotalRatingCount"] >= q) & (books["AverageRating"] > MR)]
top_books.shape
```

```python
top_books["weighted"] = top_books.apply(weighted_rating, axis=1)
above_average = top_books.sort_values("weighted", ascending=False)
above_average.shape
```
