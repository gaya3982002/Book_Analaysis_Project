```python
from category_encoders import OrdinalEncoder

maplist = [{'col': 'Rating', 
            'mapping': {"This user doesn't have any rating": 0, 'did not like it': 1,'it was ok': 2, 
                        'liked it': 3, 'really liked it': 4, 'it was amazing': 5}}]

oe = OrdinalEncoder(mapping=maplist)
```

### KMeans clustering in User ratings

```python
x = df_userratings1[['ID', 'Rating']]
```

```python
#finding optimal number of clusters using the elbow method  
from sklearn.cluster import KMeans  
wcss_list= []  #Initializing the list for the values of WCSS  

#Using for loop for iterations from 1 to 10.  
for i in range(1, 11):  
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state= 42)  
    kmeans.fit(x)  
    wcss_list.append(kmeans.inertia_)  
plt.plot(range(1, 11), wcss_list)  
plt.title('The Elobw Method Graph')  
plt.xlabel('Number of clusters(k)')  
plt.ylabel('wcss_list')  
plt.show()  
```

- k = 3

```python
#training the K-means model on a dataset  
kmeans = KMeans(n_clusters=5, init='k-means++', random_state= 42)  
y_predict= kmeans.fit_predict(x)  
```

```python
#visulaizing the clusters  
plt.scatter(x['ID'], x['Rating'])  
plt.title('Clusters of Books')  
plt.xlabel('User ID')  
plt.ylabel('Book Ratings')  
plt.legend()  
plt.show()  
```

```python
def get_user_similar_movies(user1, user2):
    common_movies = df_userratings1[df_userratings1.ID == user1].merge(df_userratings1[df_userratings1.ID == user2],
                                                                      on= "Name", how= "inner")
    return common_movies
```
