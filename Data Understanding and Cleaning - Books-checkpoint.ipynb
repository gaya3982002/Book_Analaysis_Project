**Data Description:**

This Data represents Metadata of books available on Goodreads website.

**Column Description:**
- **"ID":** Book ID on Goodreads.com.
- **"Name:** Book Title.
- **"RatingDist1:"** No of readers who gave Rating of 1 Star to the book.
- **"pagesNumber:"** Counts of Page for each book.
- **"RatingDist4:"** No of readers who gave Rating of 4 Star to the book.
- **"RatingDistTotal:"** Total No of readers who gave some Rating to the book.
- **"PublishMonth:"** Month in which book is published.
- **"PublishDay:"** Day of Month on which book is published.
- **"Publisher:"** Name of the Book Publisher.
- **"CountsOfReview:"** Count of text reviews recieved by book.
- **"PublishYear:"** Year in which book is published.
- **"Language:"** Language in which book is written.
- **"Authors:"** Name of the Authors of the book.
- **"Rating:"** Average Rating of the book.
- **"RatingDist2:"** No of readers who gave Rating of 2 Star to the book.
- **"RatingDist5:"** No of readers who gave Rating of 5 Star to the book.
- **"ISBN:"** ISBN number of the book.
- **"RatingDist3:"** No of readers who gave Rating of 3 Star to the book.
- **"Description:"** Description of Book in short.
- **"Count of text reviews:"** Count of text reviews recieved by book. 
- **"PagesNumber:"** Counts of Page for each book.

- Out of them, there are two columns which are kind of repeated viz. "pagesNumber" and "PagesNumber" also "CountsOfReview" and "Count of text reviews".


- We need more logical names for some Columns so we should make them as follows:
- "Id" : "BookID"
- "Name" : "BookTitle"
- "RatingDist1" : "1StarRatingCount"
- "pagesNumber" : "PageCount"

- "PublishMonth"
- "PublishDay"
- "Publisher"

- "PublishYear"
- "Language"
- "Authors"
- "Rating" : "AverageRating"
- "RatingDist2" : "2StarRatingCount"
- "RatingDist5" : "5StarRatingCount"
- "ISBN"
- "RatingDist3" : "3StarRatingCount"
- "Description"
- "Count of text reviews" : "TextReviewCount1"
- "PagesNumber" : "PageCount2"

```python
Books.rename(columns = {
    "Id" : "BookID",
    "Name" : "BookTitle",
    "RatingDist1" : "1StarRatingCount",
    "pagesNumber" : "PageCount",
    "RatingDist4" : "4StarRatingCount",
    "RatingDistTotal" : "TotalRatingCount",
    "CountsOfReview" : "TextReviewCount",
    "Rating" : "AverageRating",
    "RatingDist2" : "2StarRatingCount",
    "RatingDist5" : "5StarRatingCount",
    "RatingDist3" : "3StarRatingCount",
    "Count of text reviews" : "TextReviewCount1",
    "PagesNumber" : "PageCount2"
}, inplace = True)
```

```python
colnames = Books.columns
for e in colnames:
    print(e)
```
