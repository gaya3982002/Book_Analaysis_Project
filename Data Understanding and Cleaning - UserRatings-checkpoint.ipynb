# Section 2 : Data Cleaning and Data Validation

## 2.1. Data Cleaning

- During Data Inspection, we came across various problems about our data.
- Some of them were resolved by us and remaining ones will be addressed in this Section.

#### Removing the Duplicates

- We are keeping first instant of duplicated row to be unique

```python
bool_series = UserRatings.duplicated(keep='first')
UserRatings = UserRatings[~bool_series]
```

- After this the 435 duplicate rows will be deleted, so final row count should be:

362596 - 435 = 362161

```python
print(f"New Row count: {UserRatings.shape[0]}")
```

- We can conclude that we successfully removed the duplicate values

#### Remove erroneous entries

- We estalblished in above section that entries which has "BookName" == "Rating" are the records in which user did not rated any book.
- So this is unneeded data for us so we are removing those records.
- These are total 4765 records.


```python
UserRatings = UserRatings[~(UserRatings.BookName == "Rating")]
```

- So after removing these records, remaining records should be:

362161 - 4765 = 367396

```python
UserRatings.shape[0]
```

- From this we can conclude that we successfully removed these records.


- Now we need to check if these records contains all records which has "UserRatingBook" == "This user doesn't have any rating".

```python
sns.countplot(UserRatings.UserRatingBook)
plt.show()
```

- There are no values in first category now.

```python
UserRatings[UserRatings['UserRatingBook'] == "This user doesn't have any rating"]
```

```python
Ratings = UserRatings.UserRatingBook.unique()
for e in Ratings:
    print(e)
```

- Above conclusion is reinforced with above lines of code

```python
rating_dict = {'it was amazing': 5, 'really liked it': 4, 'liked it': 3, 'it was ok': 2, 'did not like it': 1}
UserRatings['UserRatingBook'] = UserRatings['UserRatingBook'].replace(rating_dict)
```
